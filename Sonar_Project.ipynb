````python
#importing the dependencies
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
````

````python
#group the features based on mean of feature for class 1 0r 0
df.groupby(60).mean().T
````

````python
#data preprocessing
#seperate x and y column
X=df.drop(columns=[60])
y=df[60]
````

````python
#TRAIN TEST SPLIT
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
````

````python
X_train.shape
````

````python
X_test.shape
````

````python
#feature scaling
scaler=StandardScaler()
````

````python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
````

````python
#model training
model=SVC()
````

````python
#train the model
model.fit(X_train_scaled,y_train)
````

````python
#model evaluation
y_train_pred=model.predict(X_train_scaled)
model_accuracy_training=classification_report(y_train,y_train_pred)
print(accuracy_score)
print("-----"*30)
print(model_accuracy_training)
````

````python
#now evaluation for predictaion
y_test_pred=model.predict(X_test_scaled)
model_accuracy_testing=classification_report(y_test,y_test_pred)
print("-----"*30)
print(model_accuracy_testing)
````

````python
#building a prediction system
def predict_object(input_features):
  #scale features
  scaled_features=scaler.transform([input_features])
  #get prediction from model
  prediction=model.predict(scaled_features)
  print(prediction)
  if prediction[0]==0:
    print("The object is a rock")
  else:
    print("The object is a mine")
````
